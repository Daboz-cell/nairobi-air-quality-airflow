```python
headers = {
"X-API-Key": API_KEY
}
 
res = requests.get(f"{BASE_URL}/locations", headers=headers)
res.json()
```

```python
NBO_Location=Latitude:  -1.286389
Longitude: 36.817223
Radius:    25 km
```

```python
params = {
    "coordinates": "-1.286389,36.817223",  # Nairobi lat,lon
    "radius": 25000,                        # 25 km, in meters
    "limit": 100
}

res = requests.get(f"{BASE_URL}/locations", headers=headers, params=params)
print(res.status_code)
data = res.json()
```

```python
results=data["results"]
results
```

```python
#Sensor=An individual air-quality sensing unit that collects environmental data
for loc in results:
    print(loc["id"], loc["name"], "->", [s["name"] for s in loc["sensors"]])
```

```python
import pandas as pd

sensor_rows = []
for loc in results:
    for s in loc["sensors"]:
        sensor_rows.append({
            "location_id": loc["id"],
            "location_name": loc["name"],
            "sensor_id": s["id"],
            "parameter": s["parameter"]["name"],
            "units": s["parameter"]["units"]
        })

sensors_df = pd.DataFrame(sensor_rows)
sensors_df
```

```python
#Measuerments = A single reported value from one sensor at one point in time
measurements_df=sensors_df["parameter"].unique()
measurements_df
```
